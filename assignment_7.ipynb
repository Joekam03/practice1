```python
outcomes = [1,2,3,4,5,6]
probabilities = [1/6] * 6
sims = 1000000
events = []
def recursive_roll():
    roll = np.random.choice(outcomes, p=probabilities)
    if roll == 6:
        return roll + recursive_roll()
    else:
        return roll
for i in range(sims):
    roll = recursive_roll()
    events.append(roll)
print(sum(events)/sims)
```

For example, try $v(y) = y^2$ or $v(y)=\sqrt{y}$ with a Bernoulli or uniform or normally distributed random variable. This can be an important thing to remember: The expectation of a transformed random variable is not the transformation of the expected value.


$$
E[X] = \int_a^b x f(x)\, dx 
     = \frac{1}{b-a} \int_a^b x\, dx
     = \frac{1}{b-a} \left[ \frac{x^2}{2} \right]_{a}^{b}
     = \frac{1}{b-a} \cdot \frac{b^2 - a^2}{2}.
$$

Using $$b^2 - a^2 = (b-a)(a+b)$$,

$$
E[X] = \frac{a+b}{2}.
$$



So the expected value of a uniform distribution is the midpoint of the bounds.

$$
\mathbb{E}[a + bX] = \int_{-\infty}^{\infty} (a + b x) \, f_X(x) \, dx
$$

$$
\int_{-\infty}^{\infty} (a + b x) f_X(x)\, dx = \int_{-\infty}^{\infty} a \, f_X(x)\, dx + \int_{-\infty}^{\infty} b x \, f_X(x)\, dx
$$

$$
= a \underbrace{\int_{-\infty}^{\infty} f_X(x)\, dx}_{=1} + b \int_{-\infty}^{\infty} x \, f_X(x)\, dx
= a + b \, \mathbb{E}[X]
$$

This works because the integral of the pdf from -inf to inf is always 1 for probabilites, meaning that that section cancels out, and you can always factor a constant out of an integral.

The bivariate normal distribution with means \(\mu_1, \mu_2\), standard deviations \(\sigma_1, \sigma_2\), and correlation \(\rho\) has pdf

$$
f_{X,Y}(x,y)
=
\frac{1}{2\pi\,\sigma_1\sigma_2\sqrt{1-\rho^2}}
\exp\!\left(
-\frac{1}{2(1-\rho^2)}
\left[
\frac{(x-\mu_1)^2}{\sigma_1^2}
- \frac{2\rho(x-\mu_1)(y-\mu_2)}{\sigma_1\sigma_2}
+ \frac{(y-\mu_2)^2}{\sigma_2^2}
\right]
\right).
$$


To show that the two are independent when rho = 0, we have to show that when 0 is plugged in for rho, this distribution is equal to $$f_{X}(x) * f_{Y}(y)$$  because when two events are independent, their joint probability is $$p_{1} * p_{2}$$

When \(\rho = 0\), the bivariate normal pdf becomes:

$$
f_{X,Y}(x,y)
=
\frac{1}{2\pi\,\sigma_1\sigma_2}
\exp\!\left(
-\frac{1}{2}
\left[
\frac{(x-\mu_1)^2}{\sigma_1^2}
+
\frac{(y-\mu_2)^2}{\sigma_2^2}
\right]
\right).
$$


Which we can see is the same as the two normal pdfs multiplied together, because the square root over the 2pi cancels out, and when you multiply two exponential terms with the same base, you would add the two exponents together, like we see here.


- For the multivariate normal, show that if $\Sigma$ is a diagonal matrix, then $X_1, X_2, ..., X_n$ are independent.


$$
f_{\mathbf{X}}(\mathbf{x})
= \frac{1}{(2\pi)^{k/2} \, |\Sigma|^{1/2}}
  \exp\!\left(
    -\tfrac{1}{2} (\mathbf{x}-\boldsymbol{\mu})^{\!\top}
    \Sigma^{-1}
    (\mathbf{x}-\boldsymbol{\mu})
  \right),
$$



First, is sigma is diagonal, then


$$
\Sigma = 
\begin{bmatrix}
\sigma_1^2 & 0 & \cdots & 0 \\
0 & \sigma_2^2 & \cdots & 0 \\
\vdots & \vdots & \ddots & \vdots \\
0 & 0 & \cdots & \sigma_k^2
\end{bmatrix}, \quad
\Sigma^{-1} = 
\begin{bmatrix}
1/\sigma_1^2 & 0 & \cdots & 0 \\
0 & 1/\sigma_2^2 & \cdots & 0 \\
\vdots & \vdots & \ddots & \vdots \\
0 & 0 & \cdots & 1/\sigma_k^2
\end{bmatrix}, \quad
|\Sigma| = \prod_{i=1}^k \sigma_i^2
$$

Because the determinant in a diagonal matrix is all the elements on the diagonal multiplied together

And the exponentiated part is equal to 



$$

$$


meaning the full pdf is 

$$
f_{\mathbf{X}}(\mathbf{x}) = \frac{1}{(2\pi)^{p/2} \prod_{i=1}^p \sigma_i}

$$

Which is the same as
$$
\prod_{i=1}^{p} \frac{1}{\sqrt{2\pi} \sigma_i}

$$


or multiplying all the pdfs together, like you would when each random variable is independent
