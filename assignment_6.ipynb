- Imagine rolling a fair single six-sided die. There are 6 outcomes, all equally likely. Derive the sample space and the space of events. What are the probabilities of the outcomes and events?

- Imagine rolling two fair six-sided die. Consider a random variable that adds up the pips on the dice. There are 11 outcomes (2 , 3, ..., 12), but not all are equally likely. Derive the sample space and **describe** the space of events. What are the probabilities of the outcomes?

When rolling a fair single six-sided dies, the sample space is {1,2,3,4,5,6}, all with probability 1/6. The space of events include the empty set, with probability 0, each individual single outcome (6 possibilities), with probability 1/6. Every possible combination of 2 outcomes (15 possibilites), each with probability 1/3. Every possible combination of 3 outcomes (20 possibilites), each with probability 1/2. Every possible combination of 4 outcomes (15 possibilites), each with probability 2/3. Every possible combination of 5 outcomes (6 possibilites), each with probabilty 5/6, and final the complete set with probability 1.

```python

outcomes = [1, 2, 3, 4, 5, 6] 
sim = 5000 
transformed_rolls = [] 
for i in range(sim): 
    roll = random.choice(outcomes) 
    roll_root = math.sqrt(roll) 
    transformed_rolls.append(roll_root) 
transformed_rolls = np.array(transformed_rolls) 

fig, axes = plt.subplots(1,2,figsize=(14,5)) 

values, counts = np.unique(transformed_rolls, return_counts=True)
probs = counts / sim
axes[0].bar(values, probs, width=0.1, color='skyblue', edgecolor='black')
axes[0].set_title('Sample PMF')

sns.ecdfplot(x=transformed_rolls, ax=axes[1], label='ecdf') 
axes[1].set_title('Sample ECDF') 

plt.show()
```

```python
roulette_wheel = {
    0: 'green', 1: 'red', 2: 'black', 3: 'red', 4: 'black', 5: 'red', 6: 'black',
    7: 'red', 8: 'black', 9: 'red', 10: 'black', 11: 'black', 12: 'red',
    13: 'black', 14: 'red', 15: 'black', 16: 'red', 17: 'black', 18: 'red',
    19: 'red', 20: 'black', 21: 'red', 22: 'black', 23: 'red', 24: 'black',
    25: 'red', 26: 'black', 27: 'red', 28: 'black', 29: 'black', 30: 'red',
    31: 'black', 32: 'red', 33: 'black', 34: 'red', 35: 'black', 36: 'red'
}
```

```python
def spin_wheel():
    return random.choice(list(roulette_wheel.keys()))
```

My possible outcomes of a roulette spin are 0-36, all with equal probability of 1/37. The means the set of events includes all possible combinations of these 37 outcomes, all with probability equal to the sum of the outcomes included within the set, or the size of the set/37. Because it would be unrealistic to list all possible outcomes, I'll instead give the probabilities of sets that correspond to some common bets.
Individual bet, ie: a set of 1 possible outcome or {0}, {1}...{36}, p = 1/37
Split, ie: a set of 2 adjacent outcomes, or {1,2},{2,3}...{35,36}, p = 2/37
Square, ie: a set of 4 adjacent outcomes, or {1,2,4,5}... p = 4/37
Line, ie: a set of 6 outcomes, or {1,2,3,4,5,6}... p = 6/37
odd, ie: a set of 18 outcomes, or {1,3,5,...35}, p = 18/37
red, ie: a set of 18 outcomes, p = 18/37

```python

def outcome(bet):
    result = spin_wheel()
    result_color = roulette_wheel[result]

    if isinstance(bet, str):
        if result_color == bet.lower():
            payout = 1
        else:
            payout = -1

    else:
        if result in bet:
            payout = 36 / len(bet) - 1
        else:
            payout = -1
    return payout

def simulate_roulette(bet, sim=1000):
    results = []
    for i in range(sim):
        payout = outcome(bet)
        results.append(payout)
    return np.array(results)

roulette_results_red = simulate_roulette('red')
roulette_results_single = simulate_roulette([7])
roulette_results_odd = simulate_roulette(list(range(1, 37, 2)))
roulette_results_split = simulate_roulette([7, 8])
roulette_results_line = simulate_roulette([1, 2, 3, 4, 5, 6])
```

```python
print(roulette_results_red.mean())
print(roulette_results_single.mean())
print(roulette_results_odd.mean())
print(roulette_results_split.mean())
print(roulette_results_line.mean())
```

As you can see, after running 1000 simulations, we see all means end up being negative. This makes sense, as any bet has the same expected value. Using the expected value formula, we know that 

E = 1/37 ((K((36/K)-1)) + (37-K) * -1)

E = 1/37( (36-K) + (K-37))

E = 1/37(36-37)
E = -1/37 
E = -0.027

This means that no matter what the bet is, over time, we expect the average payout to be -0.027.

- Logistic distribution (similar to normal): 
$$
F(x; \sigma) = \dfrac{1}{1+e^{-x/\sigma}} 
$$
with $\sigma >0$.
- Exponential distribution (similar to log-normal):
$$
F_X(x) = \begin{cases}
0, & x<0 \\
1 - e^{-\lambda x}, & x \ge 0,
\end{cases}
$$
with $\lambda > 0$.
- Negative Binomial (similar to Poisson): The probability mass function for positive integers is:
$$
f(k;r,p) = \dfrac{(k+r-1)!}{k!(r-1)!}(1-p)^k p^r, \quad \text{ for $k=0,...n$}
$$
You can interpret this as follows: Flip a coin that comes up heads with probability $p$ until you get $r$ heads, and then stop. What is the probability of stopping at each $k=0,1,2,...$?
- Categorical (similar to Bernoulli): The probability mass function over $k = 1, 2, ..., K$ categories is
$$
f(k;p_1,...,p_K) = p_1^{k=1}p_2^{k=2}...p_K^{k=K}
$$
where $0 \le p_i \le 1$ and $ \sum_{k=1}^K p_k = 1$.


## Logistic Distribution

The support of the logistic distribution is x = {-inf,inf}, as the logistic distribution always returns values in between 0 and 1, regardless of what X is. As X goes to -inf, the denominator becomes extremely large, meaning the result approaches 0 asymptotically. When X approaches positive infinity, the distribution approaches 1. 

$$
F(x; \sigma) = \dfrac{1}{1+e^{-x/\sigma}} 
$$

$$
F'(x; \sigma) = f(x; \sigma) = \dfrac{d}{dx} (1+e^{-x/\sigma})^{-1}
$$

$$
f(x; \sigma) = - (1+e^{-x/\sigma})^{-2} * -\dfrac{1}{\sigma} e^{-x/\sigma}
$$

$$
f(x; \sigma) = \dfrac{e^{-x/\sigma}}{(1+e^{-x/\sigma})^{-2} * \sigma}
$$

```python

x = np.linspace(-10, 10, 500)
sigmas = [0.5, 1, 2]
mu = 0 

fig, axes = plt.subplots(1, 2, figsize=(12, 5))
for sigma in sigmas:
    pdf = logistic.pdf(x, loc=mu, scale=sigma)
    axes[0].plot(x, pdf, label=f'σ = {sigma}')
axes[0].set_title('Logistic PDF (Density)')
axes[0].set_xlabel('x')
axes[0].set_ylabel('f(x)')
axes[0].legend()

for sigma in sigmas:
    cdf = logistic.cdf(x, loc=mu, scale=sigma)
    axes[1].plot(x, cdf, label=f'σ = {sigma}')
axes[1].set_title('Logistic CDF (Distribution)')
axes[1].set_xlabel('x')
axes[1].set_ylabel('F(x)')
axes[1].legend()

plt.tight_layout()
plt.show()

```

```python
samples = logistic.rvs(loc=mu, scale=1, size=10000)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.kdeplot(samples, ax=axes[0], label='Sample KDE', color='steelblue')
axes[0].set_title('Kernel Density Estimate (KDE)')
axes[0].set_xlabel('Value')
axes[0].set_ylabel('Density')
axes[0].legend()

sns.ecdfplot(samples, ax=axes[1], label='Sample ECDF', color='steelblue')
axes[1].set_title('Empirical Cumulative Distribution Function (ECDF)')
axes[1].set_xlabel('Value')
axes[1].set_ylabel('ECDF')
axes[1].legend()

plt.tight_layout()
plt.show()
```

As we can see, at 10,000 simulations, the KDE and ECDF appear extremely similar to the theoretical pdf and cdf we graphed above. This makes sense as the large of law numbers indicates that as the number of trials increases, our observed distribution should get closer to the theoretical distribution

## Exponential Distribution

My support in this case is all Xs greater than 0, as they are the values that produce values of greater than 0, while all other values do not

$$
F_X(x) = \begin{cases}
0, & x<0 \\
1 - e^{-\lambda x}, & x \ge 0,
\end{cases}
$$

for x < 0, 
$$F_X(x) =  0 $$ therefore $$f_X(x) =  0 $$

## Negative Binomial Distribution

My support in this case is all positive integers as each would produce values of greater than 0, while all other values do not.

```python

k = [5, 10, 20] 
p = 0.5 
r = np.arange(0, 50)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

for n in k:
    pmf = nbinom.pmf(r, n, p)
    axes[0].bar(r, pmf, alpha=0.5, label=f'n = {n}, p = {p}')
axes[0].set_title('Negative Binomial PMF')
axes[0].set_xlabel('Number of Failures')
axes[0].set_ylabel('Probability')
axes[0].legend()

for n in k:
    cdf = nbinom.cdf(r, n, p)
    axes[1].bar(r, cdf, alpha=0.5, label=f'n = {n}, p = {p}')
axes[1].set_title('Negative Binomial CDF')
axes[1].set_xlabel('Number of Failures')
axes[1].set_ylabel('Cumulative Probability')
axes[1].legend()

plt.tight_layout()
plt.show()
```

```python

k, p = 5, 0.5 
samples = nbinom.rvs(k, p, size=10000)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

sns.histplot(samples, bins=np.arange(samples.min(), samples.max()+2)-0.5, 
             stat='probability', discrete=True, ax=axes[0], color='steelblue', label='Sample PMF')
axes[0].set_title('Negative Binomial PMF (Histogram)')
axes[0].set_xlabel('Number of Failures')
axes[0].set_ylabel('Probability')
axes[0].legend()

sns.ecdfplot(samples, ax=axes[1], label='Sample ECDF', color='steelblue')
axes[1].set_title('Negative Binomial ECDF')
axes[1].set_xlabel('Number of Failures')
axes[1].set_ylabel('ECDF')
axes[1].legend()

plt.tight_layout()
plt.show()
```

```python

categories = ['red', 'green', 'yellow', 'blue', 'orange']
probabilities = [0.2, 0.1, 0.3, 0.25, 0.15]

pmf = probabilities

cdf = np.cumsum(probabilities)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))


axes[0].bar(categories, pmf, color=categories)
axes[0].set_title('Categorical PMF')
axes[0].set_xlabel('Category')
axes[0].set_ylabel('Probability')


axes[1].step(categories, cdf, where='mid', color='black')
axes[1].scatter(categories, cdf, color=categories)
axes[1].set_title('Categorical CDF')
axes[1].set_xlabel('Category')
axes[1].set_ylabel('Cumulative Probability')

plt.tight_layout()
plt.show()
```

```python
n_sims = 10000

samples = np.random.choice(categories, size=n_sims, p=probabilities)

pmf_counts = [np.sum(samples == cat) for cat in categories]
pmf_empirical = np.array(pmf_counts) / n_sims

cdf_empirical = np.cumsum(pmf_empirical)

fig, axes = plt.subplots(1, 2, figsize=(12, 5))

axes[0].bar(categories, pmf_empirical, color=categories)
axes[0].set_title('Empirical PMF from 10,000 Simulations')
axes[0].set_xlabel('Category')
axes[0].set_ylabel('Probability')

axes[1].step(categories, cdf_empirical, where='mid', color='black')
axes[1].scatter(categories, cdf_empirical, color=categories)
axes[1].set_title('Empirical CDF from 10,000 Simulations')
axes[1].set_xlabel('Category')
axes[1].set_ylabel('Cumulative Probability')

plt.tight_layout()
plt.show()
```

Our categorical distribution appears to match the best out of all. The theoretical and observed are almost exactly the same.
